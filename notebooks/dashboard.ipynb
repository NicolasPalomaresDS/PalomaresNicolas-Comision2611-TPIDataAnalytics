# Análisis de Ventas, Marketing y Clientes

## Introducción

El presente análisis integra tres fuentes de datos correspondientes al año 2024: un dataset de ventas con 3.000 transacciones de 30 productos distribuidos en tres categorías (Decoración, Electrodomésticos y Electrónica), un dataset de marketing con 90 campañas distribuidas en tres canales (Email, RRSS y TV), y un dataset de 567 clientes con información demográfica y de ingresos.

El proceso incluyó limpieza y transformación de datos, análisis exploratorio con visualizaciones, ingeniería de variables y entrenamiento de un modelo de clasificación con XGBoost. Los resultados se presentan a través de dos dashboards interactivos: uno enfocado en ventas y marketing, y otro en el perfil de clientes.

## Resumen Ejecutivo

Análisis exploratorio de tres datasets (ventas, marketing y clientes) con el objetivo de identificar patrones de comportamiento, evaluar el rendimiento de campañas y construir un modelo predictivo.

---

## Principales Hallazgos

### Ventas y Marketing

- **Estacionalidad:** Mayo fue el mes de mayor facturación (144.000) y junio el de menor (107.000). La segunda mitad del año mostró ventas más estables pero por debajo del primer semestre.
- **Categorías equilibradas:** Decoración, Electrodomésticos y Electrónica presentan volúmenes de venta muy similares, sin una categoría dominante.
- **Productos destacados:** Lámpara de mesa, Auriculares y Microondas lideran en ventas totales. Lámpara de mesa también lidera en ratio ventas/costo.
- **Eficiencia por canal:** Email presenta el mayor volumen de ventas con el menor costo de campaña. No se encontró correlación entre inversión en marketing y ventas por producto.

### Clientes

- **Perfil etario:** La mayoría de los clientes tienen entre 25 y 45 años. Muy pocos clientes mayores de 65 años.
- **Ingresos homogéneos:** Los ingresos promedio por ciudad son muy similares entre sí (32.000 - 37.000), sin diferencias significativas entre regiones.
- **Outliers de ingresos:** Se identificaron valores atípicos bajos (menores a $1.000) reemplazados por la mediana, y valores altos conservados por considerarse plausibles.

### Modelo Predictivo

- Se entrenó un modelo XGBoost para predecir la categoría del producto a partir de variables transaccionales y de campaña.
- El modelo alcanzó un **F1-score de 0.96**, siendo `duracion_campanha` la variable más determinante. Sin embargo, el modelo no se considera apto para producción, ya que es posible que las métricas logradas sean por un posible data leakage.
- Los intentos de predecir variables continuas (`cantidad`, `precio`) resultaron en R² negativo, consistente con la ausencia de correlaciones en el dataset.
